```python
from colorama import Fore, Style
def obtener_datos_entrada():
    
    ID= input("Introduzca los ID para vizualizar (separado por coma): ").split(",")
    
    for i,v in enumerate(ID):
        ID[i]=int(ID[i]) #el tipo es número entero
    return list(ID) #Set es para no aparescan numeros dublicados

```

```python
def desplegar_pantalla_inicio():
    lineac = 50 * chr(0x2015)
    
    print(lineac)
    print("Sistema XYZ".center(50))
    print(lineac)

    # Se leen los datos de la pantalla
    ID = obtener_datos_entrada()
    
    print(lineac)
    
    return ID
```

```python
def transformar_datos(r, e):
    dicc = {}# Diccionario vacío para almancenar la información

    # Recorremos el renglón y extraemos el key y el value
    for k, v in zip(e, r.split(',')): # se utiliza para combinar dos o más iterables, en un solo 
        dicc[k] = v # Agregamos la info al diccionario

    # Corregimos la entrada 'TA' (para quitarle el cambio de línea)
    dicc['TA'] = v[:-1]

    # Regresamos el diccionario con la información
    return dicc
```

```python
def calcular_nivel_IMC(imc):
    if 18.5 > imc:
        nivel = 'N1'
    elif (18.5 <= imc <=24.9):
        nivel = 'N2'
    elif (25 <= imc <= 29.9):
        nivel = 'N3'
    else: #arriba de 30
        nivel = 'N4'
    return nivel
```

```python
def desplegar_datos(d):
    # Diccionario para los textos
    dicc_e = dict(Pa = "Paciente", S = "Sexo",
                  E = "Edad",  Pe = "Peso",
                  A = "Estatura",
                  IMC = "IMC",
                  Di = "Diabetes",
                  H = "Hipertensión",
                  De = "Depresión",
                  TS = "Trastorno del sueño",
                  TA = "Trastorno de la alimentación"
                 )  
    for k, v in d.items():
        # Checamos el IMC
        v_imc = ''
        if k == 'IMC':
            v_imc = calcular_nivel_IMC(float(v))

        # Checamos si hay alguna alerta de salud
        if v == 'Si' or v_imc == 'N4':            
            print(Fore.RED + Style.BRIGHT + "{:>30s} : {:^5s}".format(dicc_e[k], str(v)) +  Style.RESET_ALL)
        else:
            print("{:>30s} : {:^5s}".format(dicc_e[k], str(v)))
```

```python
def abrir_archivo(arc = "./personal_data.csv"):
    f = open(arc)
    # Extraemos encabezado en forma de lista
    e = f.readline().split(',')
    # Eliminamos el '\n' (el último caracter) del último elemento
    e[-1] = e[-1][:-1]
    # Regresamos el descriptor del archivo y el encabezado
    return f, e
```

```python
def cerrar_archivo(file_descriptor):
    file_descriptor.close()
```

```python
def leer_archivo(IDs):
    lineac = 50 * chr(0x2015)
    f, encabezado = abrir_archivo()# Abrimos el archivo y obtenemos el encabezado

    # Construimos una lista vacía para almacenar 
    IDs_lista = []# la información de los IDs solicitados

    for i, r in enumerate(f, start=1): #,r

        # Checamos si el id está en la lista de los solicitados
        if i in IDs:

            # En caso de que si sea uno de los solicitados
            # vaciamos la información en un diccionario
            dicc = transformar_datos(r, encabezado)
        
            # Presentamos la información en la pantalla
            # con el formato requerido
            desplegar_datos(dicc)

            print(lineac)
     # Cerramos el archivo
    cerrar_archivo(f)
```

```python
def ejecutar():
    leer_archivo(desplegar_pantalla_inicio())
if __name__ == '__main__':
    ejecutar()
```
